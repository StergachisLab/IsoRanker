# IsoSeq

IsoSeq_smk already tags the FLNC BAMs with the isoform assigned to each read. This is the "in" tag. However, the following
code can be used to tag the BAMs using the read stat file from IsoSeq:

```python
import pysam
import gzip
import pandas as pd
import glob
import os

# I have bam files in my current directory. I also have a file called OUT.transcript_model_reads.tsv.gz with two columns: 
# #read_id and transcript_id. I want to add the transcript_id from OUT.transcript_model_reads.tsv.gz as a new tag to the 
# reads in the bam files. 

# Load read => transcript map
map_df = pd.read_csv("/mmfs1/gscratch/stergachislab/asedeno/data/github/IsoSeq_smk/results/collapse/collapsed.read_stat.txt", sep="\t")
read_to_tx = dict(zip(map_df["id"], map_df["pbid"]))

# Process all BAM files in current directory
for bam_path in glob.glob("*.tagged.bam"):
    print(f"Processing {bam_path} ...")

    in_bam = pysam.AlignmentFile(bam_path, "rb")
    out_bam_path = os.path.splitext(bam_path)[0] + ".newtagged.bam"
    out_bai_path = out_bam_path + ".bai"
    out_bam = pysam.AlignmentFile(out_bam_path, "wb", header=in_bam.header)

    for read in in_bam:
        rid = read.query_name
        if rid in read_to_tx:
            read.set_tag("in", read_to_tx[rid], value_type="Z")  # string tag
        out_bam.write(read)

    in_bam.close()
    out_bam.close()
    print(f"→ Saved: {out_bam_path}")

    # Create BAM index
    print(f"Indexing {out_bam_path} ...")
    pysam.index(out_bam_path, out_bai_path)
    print(f"→ Saved: {out_bam_path}")
    print(f"→ Indexed: {out_bai_path}\n")
```
