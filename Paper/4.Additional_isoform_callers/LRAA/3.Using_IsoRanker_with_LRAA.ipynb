# Making read stat file

```python
import pandas as pd

df = pd.read_csv("/mmfs1/gscratch/stergachislab/asedeno/data/LRAA/20250922_095432_LRAA_wf/out/mergedQuantTracking/all.LRAA.quant.tracking", 
                 sep='\t') # Specify the compression type

df_duplicates_removed = df.loc[df.groupby("read_name")["frac_assigned"].idxmax()].reset_index(drop=True)

# Keep only the desired columns
df_duplicates_removed_reformatted = df_duplicates_removed[["read_name", "transcript_id"]]

# Rename columns
df_duplicates_removed_reformatted = df_duplicates_removed_reformatted.rename(columns={"read_name": "id", "transcript_id": "pbid"})

# Ensure column order (id first, pbid second)
df_duplicates_removed_reformatted = df_duplicates_removed_reformatted[["id", "pbid"]]

df_duplicates_removed_reformatted.to_csv("LRAA_read_stat.txt", sep="\t", index=False)
```
